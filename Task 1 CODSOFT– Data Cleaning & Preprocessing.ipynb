```python
import pandas as pd

file_path = r"C:\Users\santo\Downloads\meesho_product_dataset_india_30.xlsx"

df = pd.read_excel(file_path)

print(df.head())
print(df.shape)
print(df.columns)
```

```python
# Check number of rows and columns
print("Shape of dataset:", df.shape)

# Check column names
print("\nColumn names:")
print(df.columns)

# Check data types
print("\nData types:")
print(df.dtypes)

# View first 5 rows
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
missing = df.isnull().sum()

missing_percentage = (missing / len(df)) * 100

print("Missing Values:")
print(missing)

print("\nMissing Value Percentage:")
print(missing_percentage)
```

```python
# Check total number of duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Display duplicate records
duplicates = df[df.duplicated()]

print(duplicates)
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Check unique values in categorical columns

print("Categories:")
print(df["Category"].unique())

print("\nSub-Categories:")
print(df["Sub-Category"].unique())

print("\nAvailability:")
print(df["Availability"].unique())

print("\nSeller Names:")
print(df["Seller Name"].unique())
```

```python
# Remove unnecessary spaces from text columns

text_columns = ["Product Name", "Category", "Sub-Category",
                "Seller Name", "Availability"]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Extra spaces removed successfully.")
```

```python
# Check current data types
print(df.dtypes)
```

```python
# Convert numerical columns to numeric data types
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Discount Price"] = pd.to_numeric(df["Discount Price"], errors="coerce")
df["Ratings"] = pd.to_numeric(df["Ratings"], errors="coerce")
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

# Check data types again
print(df.dtypes)
```

```python
print(df[["Price", "Discount Price", "Ratings", "Reviews"]].head())
```

```python
print(df.isnull().sum())
```

```python
numeric_columns = ["Price", "Discount Price", "Ratings", "Reviews"]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
text_columns = [
    "Product Name",
    "Category",
    "Sub-Category",
    "Seller Name",
    "Delivery Time",
    "Availability"
]

for col in text_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
    
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Final dataset information
print("Final Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nFirst 5 Rows:")
print(df.head())
```

```python
output_path = r"C:\Users\santo\Downloads\meesho_product_dataset_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```
