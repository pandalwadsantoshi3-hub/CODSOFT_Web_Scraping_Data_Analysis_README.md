```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
file_path = r"C:\Users\santo\Downloads\meesho_product_dataset_cleaned.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
category_counts = df["Category"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(category_counts.index, category_counts.values)

plt.title("Number of Products by Category")
plt.xlabel("Category")
plt.ylabel("Number of Products")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

LINE CHART

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Product Name"], df["Price"], marker="o", label="Original Price")
plt.plot(df["Product Name"], df["Discount Price"], marker="o", label="Discount Price")

plt.title("Original Price vs Discount Price")
plt.xlabel("Product Name")
plt.ylabel("Price (₹)")

plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(14, 6))

plt.plot(df.index, df["Price"], marker="o", label="Original Price")
plt.plot(df.index, df["Discount Price"], marker="o", label="Discount Price")

plt.title("Original Price vs Discount Price")
plt.xlabel("Product Index")
plt.ylabel("Price (₹)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
availability_counts = df["Availability"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    availability_counts.values,
    labels=availability_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Availability")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["Price"],
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Product Prices")
plt.xlabel("Price (₹)")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Price"],
    df["Ratings"],
    alpha=0.7
)

plt.title("Price vs Ratings")
plt.xlabel("Price (₹)")
plt.ylabel("Ratings")

plt.tight_layout()
plt.show()
```

```python
top_10 = df.nlargest(10, "Reviews")

plt.figure(figsize=(10,6))

plt.barh(
    top_10["Product Name"],
    top_10["Reviews"]
)

plt.title("Top 10 Most Reviewed Products")
plt.xlabel("Number of Reviews")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

```python
avg_rating = df.groupby("Category")["Ratings"].mean()

plt.figure(figsize=(8,5))

plt.bar(
    avg_rating.index,
    avg_rating.values
)

plt.title("Average Rating by Category")
plt.xlabel("Category")
plt.ylabel("Average Rating")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

correlation = df[
    ["Price", "Discount Price", "Ratings", "Reviews"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
total_products = len(df)
average_price = df["Price"].mean()
average_rating = df["Ratings"].mean()
total_reviews = df["Reviews"].sum()

print("====================================")
print("       MEESHO PRODUCT DASHBOARD")
print("====================================")

print("Total Products:", total_products)
print("Average Price: ₹", round(average_price, 2))
print("Average Rating:", round(average_rating, 2))
print("Total Reviews:", int(total_reviews))

print("====================================")
```
