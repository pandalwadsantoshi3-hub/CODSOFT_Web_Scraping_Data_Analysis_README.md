```python
import pandas as pd

file_path = r"C:\Users\santo\Downloads\customers-100.xlsx"

df = pd.read_excel(file_path)

print(df.head())
print("\nShape:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Dataset information
print("\nDataset Information:")
df.info()
```

```python
country_counts = df["Country"].value_counts()

print("Customers by Country:")
print(country_counts)
```

```python
city_counts = df["City"].value_counts()

print("Customers by City:")
print(city_counts.head(10))
```

```python
print("Total Customers:", df["Customer Id"].nunique())
print("Total Countries:", df["Country"].nunique())
print("Total Cities:", df["City"].nunique())
print("Total Companies:", df["Company"].nunique())
```

```python
import matplotlib.pyplot as plt

country_counts = df["Country"].value_counts().head(10)

plt.figure(figsize=(10, 6))
country_counts.plot(kind="bar")

plt.title("Top 10 Countries by Number of Customers")
plt.xlabel("Country")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

city_counts = df["City"].value_counts().head(10)

plt.figure(figsize=(10, 6))
city_counts.plot(kind="bar")

plt.title("Top 10 Cities by Number of Customers")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Check Subscription Date
print(df["Subscription Date"].head())
print(df["Subscription Date"].dtype)
```

```python
# Extract year and month
df["Subscription Year"] = df["Subscription Date"].dt.year
df["Subscription Month"] = df["Subscription Date"].dt.month_name()

# Count customers by subscription year
year_counts = df["Subscription Year"].value_counts().sort_index()

print("Customers by Subscription Year:")
print(year_counts)
```

```python
import matplotlib.pyplot as plt

year_counts = df["Subscription Year"].value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.plot(year_counts.index, year_counts.values, marker="o")

plt.title("Customer Subscription Trend by Year")
plt.xlabel("Subscription Year")
plt.ylabel("Number of Customers")
plt.grid(True)
plt.tight_layout()

plt.show()
```
